## Sección 1: Construyendo Grafos

La representación más directa de un grafo en Python es un **diccionario de listas de adyacencia**:
cada clave es un vértice y su valor es la lista de vértices vecinos.

Para un grafo **no dirigido** con arista $\{u, v\}$, la representamos dos veces: $u \to v$ y $v \to u$.
Para un **dígrafo**, cada arista $(u, v)$ aparece solo en la lista de $u$.

### Ventajas del diccionario de adyacencia
- Iterar sobre vecinos de $u$: $O(\deg(u))$ — solo recorremos los vecinos reales
- Agregar un vértice o arista: $O(1)$ amortizado
- Espacio: $O(V + E)$ — eficiente para grafos dispersos


````python
# Grafo no dirigido: 6 nodos, estilo lista de adyacencia
grafo_nd = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E'],
}

# Construir el mismo grafo con NetworkX para visualizarlo
G_nd = nx.Graph()
for u, vecinos in grafo_nd.items():
    for v in vecinos:
        G_nd.add_edge(u, v)

fig, ax = plt.subplots(figsize=(7, 5))
pos = nx.spring_layout(G_nd, seed=7)
nx.draw_networkx(G_nd, pos=pos, ax=ax,
                 node_color=COLORS["node"], node_size=800,
                 edge_color=COLORS["edge"], width=2,
                 font_size=13, font_weight='bold')
ax.set_title("Grafo no dirigido — 6 vértices, 5 aristas", fontsize=14)
ax.axis('off')
plt.tight_layout()
plt.show()

print(f"Vértices: {list(G_nd.nodes())}")
print(f"Aristas:  {list(G_nd.edges())}")
print(f"|V| = {G_nd.number_of_nodes()}, |E| = {G_nd.number_of_edges()}")
````

````python
# Dígrafo: 5 nodos
grafo_dir = {
    'S': ['A', 'B'],
    'A': ['C'],
    'B': ['C', 'D'],
    'C': ['D'],
    'D': [],
}

G_dir = nx.DiGraph()
for u, vecinos in grafo_dir.items():
    for v in vecinos:
        G_dir.add_edge(u, v)

fig, ax = plt.subplots(figsize=(7, 5))
pos = nx.spring_layout(G_dir, seed=42)
nx.draw_networkx(G_dir, pos=pos, ax=ax,
                 node_color=COLORS["accent"], node_size=800,
                 edge_color=COLORS["edge"], width=2,
                 arrows=True, arrowsize=20,
                 font_size=13, font_weight='bold')
ax.set_title("Dígrafo — 5 vértices, 6 aristas", fontsize=14)
ax.axis('off')
plt.tight_layout()
plt.show()

print(f"Vértices: {list(G_dir.nodes())}")
print(f"Aristas:  {list(G_dir.edges())}")
````

## Sección 2: Matrices de Adyacencia

La **matriz de adyacencia** $A \in \{0,1\}^{V \times V}$ define:

$$A[i][j] = \begin{cases} 1 & \text{si } (i,j) \in E \\ 0 & \text{si no} \end{cases}$$

Para grafos no dirigidos, $A$ es **simétrica**: $A[i][j] = A[j][i]$.

### Complejidades
| Operación | Lista de adyacencia | Matriz de adyacencia |
|-----------|--------------------|-----------------------|
| ¿Existe arista $(u,v)$? | $O(\deg(u))$ | $O(1)$ |
| Vecinos de $u$ | $O(\deg(u))$ | $O(V)$ |
| Espacio | $O(V + E)$ | $O(V^2)$ |

La matriz es preferible cuando $E \approx V^2$ (grafos **densos**); la lista lo es cuando $E \ll V^2$ (grafos **dispersos**).


````python
# Convertir diccionario de adyacencia a matriz numpy
def dict_a_matriz(grafo):
    nodos = sorted(grafo.keys())
    idx = {n: i for i, n in enumerate(nodos)}
    n = len(nodos)
    A = np.zeros((n, n), dtype=int)
    for u, vecinos in grafo.items():
        for v in vecinos:
            A[idx[u], idx[v]] = 1
    return A, nodos

A, nodos = dict_a_matriz(grafo_nd)
print("Matriz de adyacencia del grafo no dirigido:")
print("Nodos:", nodos)
print(A)
print(f"\nEs simétrica: {np.array_equal(A, A.T)}")

# Visualizar la matriz
fig, ax = plt.subplots(figsize=(5, 4))
im = ax.imshow(A, cmap='Blues', vmin=0, vmax=1)
ax.set_xticks(range(len(nodos))); ax.set_xticklabels(nodos)
ax.set_yticks(range(len(nodos))); ax.set_yticklabels(nodos)
for i in range(len(nodos)):
    for j in range(len(nodos)):
        ax.text(j, i, str(A[i,j]), ha='center', va='center', fontsize=12,
                color='white' if A[i,j] else 'black')
ax.set_title("Matriz de adyacencia $A$", fontsize=13)
plt.tight_layout()
plt.show()
````

````python
# Verificar complejidades
idx = {n: i for i, n in enumerate(nodos)}

# Consulta de arista: O(1) con matriz
u, v = 'A', 'B'
existe = bool(A[idx[u], idx[v]])
print(f"¿Existe arista ({u},{v})? {existe}  ← O(1) con matriz")

# Iterar vecinos: O(V) con matriz (recorremos toda la fila)
fila_A = A[idx['A']]
vecinos_de_A_matriz = [nodos[j] for j, val in enumerate(fila_A) if val]
vecinos_de_A_lista  = grafo_nd['A']
print(f"Vecinos de A (matriz, O(V)={len(nodos)}): {vecinos_de_A_matriz}")
print(f"Vecinos de A (lista,  O(deg)={len(vecinos_de_A_lista)}): {vecinos_de_A_lista}")
print("\n→ Con la lista de adyacencia solo tocamos los vecinos reales.")
print("  Con la matriz recorremos TODOS los nodos aunque la mayoría sean 0.")
````

## Sección 3: Grado de Vértices

El **grado** $\deg(v)$ de un vértice es el número de aristas incidentes.

- En un grafo no dirigido: $\deg(v) = |\{u : \{u,v\} \in E\}|$
- En un dígrafo distinguimos **grado de entrada** $\deg^-(v)$ y **grado de salida** $\deg^+(v)$

El **lema del apretón de manos** establece:
$$\sum_{v \in V} \deg(v) = 2|E|$$

Esto se debe a que cada arista contribuye 2 al total (una por cada extremo).


````python
# Grado en grafo no dirigido
print("=== Grado en grafo no dirigido ===")
for nodo, vecinos in sorted(grafo_nd.items()):
    print(f"  deg({nodo}) = {len(vecinos)}")
total_grado = sum(len(v) for v in grafo_nd.values())
num_aristas = total_grado // 2
print(f"\nΣ deg = {total_grado} = 2 × {num_aristas} aristas  ✓")

print("\n=== Grado en y salida en dígrafo ===")
for nodo in sorted(grafo_dir.keys()):
    out_deg = len(grafo_dir[nodo])
    in_deg  = sum(1 for u in grafo_dir for v in grafo_dir[u] if v == nodo)
    print(f"  {nodo}: deg⁺ (salida) = {out_deg},  deg⁻ (entrada) = {in_deg}")
````

## Sección 4: Componentes Conexas

Una **componente conexa** de un grafo no dirigido es un subconjunto maximal $C \subseteq V$ tal que
para todo par $u, v \in C$ existe un camino de $u$ a $v$.

Para encontrarlas, usamos una variante de **DFS** (Depth-First Search):
1. Marcar todos los vértices como no visitados
2. Para cada vértice no visitado, hacer DFS desde él — todo lo alcanzado forma una componente

La complejidad es $O(V + E)$ — visitamos cada vértice y cada arista exactamente una vez.


````python
# Grafo desconectado con 3 componentes
grafo_desc = {
    'A': ['B', 'C'],  # componente 1
    'B': ['A'],
    'C': ['A'],
    'D': ['E'],       # componente 2
    'E': ['D', 'F'],
    'F': ['E'],
    'G': [],          # componente 3 (nodo aislado)
}

def dfs_componente(grafo, inicio, visitados):
    """DFS iterativo que devuelve todos los nodos alcanzables desde 'inicio'."""
    componente = []
    pila = [inicio]
    while pila:
        nodo = pila.pop()
        if nodo in visitados:
            continue
        visitados.add(nodo)
        componente.append(nodo)
        for vecino in grafo[nodo]:
            if vecino not in visitados:
                pila.append(vecino)
    return componente

def componentes_conexas(grafo):
    """Devuelve una lista de componentes (cada componente es una lista de nodos)."""
    visitados = set()
    componentes = []
    for nodo in grafo:
        if nodo not in visitados:
            comp = dfs_componente(grafo, nodo, visitados)
            componentes.append(sorted(comp))
    return componentes

comps = componentes_conexas(grafo_desc)
print(f"Número de componentes conexas: {len(comps)}")
for i, c in enumerate(comps, 1):
    print(f"  Componente {i}: {c}")
````

````python
# Visualizar el grafo desconectado
G_desc = nx.Graph()
for u, vecinos in grafo_desc.items():
    G_desc.add_node(u)
    for v in vecinos:
        G_desc.add_edge(u, v)

comp_colors = {}
palette = [COLORS["primary"], COLORS["secondary"], COLORS["accent"]]
for i, comp in enumerate(comps):
    for nodo in comp:
        comp_colors[nodo] = palette[i]

node_colors = [comp_colors[n] for n in G_desc.nodes()]

fig, ax = plt.subplots(figsize=(8, 5))
pos = nx.spring_layout(G_desc, seed=5)
nx.draw_networkx(G_desc, pos=pos, ax=ax,
                 node_color=node_colors, node_size=900,
                 edge_color=COLORS["edge"], width=2,
                 font_size=13, font_weight='bold', font_color='white')
ax.set_title("Grafo con 3 componentes conexas (colores distintos)", fontsize=13)
ax.axis('off')
plt.tight_layout()
plt.show()
````

## Sección 5: Representación vs. Densidad

La **densidad** de un grafo no dirigido se define como:

$$\rho = \frac{2|E|}{|V|(|V|-1)}$$

- $\rho \approx 0$: grafo **disperso** (sparse) — la lista de adyacencia es eficiente en memoria
- $\rho \approx 1$: grafo **denso** (dense) — la matriz de adyacencia ocupa similar espacio pero da consultas $O(1)$

En algoritmos de búsqueda (BFS, DFS), el coste total es $O(V + E)$:
- Grafo disperso con $E = O(V)$: costo $O(V)$
- Grafo denso con $E = O(V^2)$: costo $O(V^2)$

Elegir bien la representación **no cambia la complejidad asintótica** pero sí el factor constante y el uso de memoria.


````python
# Grafo disperso: 100 nodos, ~120 aristas
G_sparse = nx.gnm_random_graph(n=100, m=120, seed=1)

# Grafo denso: 20 nodos, ~380 aristas (de 190 posibles → casi completo)
G_dense = nx.gnm_random_graph(n=20, m=180, seed=2) #modifique de 190 a 180 para que no sea completo, pero sigue siendo denso

def densidad(G):
    n = G.number_of_nodes()
    e = G.number_of_edges()
    return 2*e / (n*(n-1)) if n > 1 else 0

print("Grafo disperso:")
print(f"  |V| = {G_sparse.number_of_nodes()}, |E| = {G_sparse.number_of_edges()}")
print(f"  Densidad ρ = {densidad(G_sparse):.4f}")

print("\nGrafo denso:")
print(f"  |V| = {G_dense.number_of_nodes()}, |E| = {G_dense.number_of_edges()}")
print(f"  Densidad ρ = {densidad(G_dense):.4f}")
````

````python
# Comparar memoria: lista de adyacencia vs matriz
def memoria_lista(G):
    """Bytes estimados de la lista de adyacencia."""
    total = 0
    for nodo in G.nodes():
        total += sys.getsizeof(list(G.neighbors(nodo)))
    return total

def memoria_matriz(n):
    """Bytes de una matriz numpy n×n de enteros."""
    return np.zeros((n, n), dtype=np.int8).nbytes

print("=== Uso de memoria ===")
for nombre, G in [("Disperso (100n, 120e)", G_sparse), ("Denso (20n, 190e)", G_dense)]:
    n = G.number_of_nodes()
    ml = memoria_lista(G)
    mm = memoria_matriz(n)
    print(f"\n{nombre}:")
    print(f"  Lista de adyacencia: {ml:6} bytes")
    print(f"  Matriz de adyacencia:{mm:6} bytes")
    print(f"  Razón matriz/lista:  {mm/ml:.2f}x")

print("\n→ Para el grafo disperso, la matriz ocupa MUCHO más memoria.")
print("  Para el grafo denso, la diferencia se achica.")
````

````python
# Comparar tiempo de consulta de arista
# Construir ambas representaciones para G_sparse
nodos_s = list(G_sparse.nodes())
idx_s = {n: i for i, n in enumerate(nodos_s)}
A_sparse = nx.to_numpy_array(G_sparse, dtype=np.int8)
adj_list_sparse = {n: set(G_sparse.neighbors(n)) for n in nodos_s}

N_consultas = 100_000
pares = [(nodos_s[i % len(nodos_s)], nodos_s[(i*7+3) % len(nodos_s)])
         for i in range(N_consultas)]

# Tiempo con matriz
t0 = time.perf_counter()
for u, v in pares:
    _ = A_sparse[idx_s[u], idx_s[v]]
t_matriz = time.perf_counter() - t0

# Tiempo con lista
t0 = time.perf_counter()
for u, v in pares:
    _ = v in adj_list_sparse[u]
t_lista = time.perf_counter() - t0

print(f"Consulta de arista ({N_consultas:,} veces) — Grafo disperso (100 nodos):")
print(f"  Matriz numpy:        {t_matriz*1000:.2f} ms")
print(f"  Lista (set lookup):  {t_lista*1000:.2f} ms")
print()
print("Nota: con sets Python, la consulta en lista también es O(1) amortizado.")
print("La diferencia real aparece al *iterar* vecinos, no al consultar aristas.")
````

```python
vocab = ['GATO', 'DATO', 'PATO', 'RATO', 'RATA', 'DATA', 'RUTA', 'MUTA', 'MULA', 'BALA']
```

````python
def difieren_en_una(palabra1, palabra2):
    """Función auxiliar para saber si dos palabras difieren en exactamente 1 letra."""
    diferencias = 0
    for letra1, letra2 in zip(palabra1, palabra2):
        if letra1 != letra2:
            diferencias += 1
            if diferencias > 1: # Optimización: si ya hay más de 1, rompemos el ciclo
                return False
    return diferencias == 1

def construir_grafo_palabras(vocab):
    """Construye el grafo usando una lista de adyacencia."""
    # Inicializamos el diccionario con listas vacías para cada nodo
    grafo = {palabra: [] for palabra in vocab}
    
    # Comparamos cada par de palabras (evitando comparar una palabra consigo misma
    # o comparar el mismo par dos veces)
    for i in range(len(vocab)):
        for j in range(i + 1, len(vocab)):
            if difieren_en_una(vocab[i], vocab[j]):
                # Como es un grafo NO dirigido, añadimos la arista en ambas direcciones
                grafo[vocab[i]].append(vocab[j])
                grafo[vocab[j]].append(vocab[i])
                
    return grafo

# Ejecutamos el paso 1
vocab = ['GATO', 'DATO', 'PATO', 'RATO', 'RATA', 'DATA', 'RUTA', 'MUTA', 'MULA', 'BALA']
grafo = construir_grafo_palabras(vocab)
````

````python
def obtener_mayor_grado(grafo):
    # 1. Encontramos cuál es el número máximo de conexiones que tiene cualquier nodo
    max_grado = max(len(vecinos) for vecinos in grafo.values())
    
    # 2. Filtramos el diccionario para quedarnos solo con las palabras que tengan ese grado
    palabras_top = [nodo for nodo, vecinos in grafo.items() if len(vecinos) == max_grado]
    
    return max_grado, palabras_top

grado_max, palabras_populares = obtener_mayor_grado(grafo)
print(f"El grado máximo es {grado_max} y pertenece a: {palabras_populares}")
````

````python
def contar_componentes_conexas(grafo):
    visitados = set()
    num_componentes = 0
    
    for nodo in grafo:
        # Si el nodo no está en visitados, hemos encontrado una nueva "isla"
        if nodo not in visitados:
            num_componentes += 1
            
            pila = [nodo]
            
            while pila:
                actual = pila.pop() # Saca el más reciente
                
                if actual not in visitados:
                    visitados.add(actual)
                    
                    # Añadimos los vecinos a la frontera (pila)
                    for vecino in grafo[actual]:
                        if vecino not in visitados:
                            pila.append(vecino)
            
    return num_componentes

total_componentes = contar_componentes_conexas(grafo)
print(f"El grafo tiene {total_componentes} componentes conexas.")
````


### Ejercicio 2: Teorema de Euler

Un grafo no dirigido tiene un **camino euleriano** (que recorre cada arista exactamente una vez) si y solo si tiene exactamente 0 o 2 vértices de grado impar.

Escribe una función `tiene_camino_euleriano(grafo)` que verifique esta condición. Pruébala con:
- El grafo no dirigido de la Sección 1
- Un ciclo de 5 nodos
- El grafo de los Puentes de Königsberg (el grafo clásico que Euler estudió)

```python
# Puentes de Königsberg: 4 nodos (A=tierra norte, B=isla, C=tierra sur, D=tierra este)
konigsberg = {
    'A': ['B', 'B', 'D'],     # dos puentes a B, uno a D
    'B': ['A', 'A', 'C', 'C', 'D'],
    'C': ['B', 'B', 'D'],
    'D': ['A', 'B', 'C'],
}
# Tu código aquí
```


````python
def tiene_camino_euleriano(grafo):
    """
    Verifica si un grafo no dirigido tiene un camino euleriano.
    Condición: Exactamente 0 o 2 vértices deben tener un grado impar.
    """
    nodos_impares = 0
    
    # 1. Recorremos cada nodo y su lista de vecinos en el grafo
    for nodo, vecinos in grafo.items():
        
        # 2. El grado de un nodo es simplemente la cantidad de vecinos que tiene
        grado = len(vecinos)
        
        # 3. Verificamos si el grado es impar (el residuo de la división entre 2 no es 0)
        if grado % 2 != 0:
            nodos_impares += 1
            
    # 4. Aplicamos el Teorema de Euler
    if nodos_impares == 0 or nodos_impares == 2:
        return True
    else:
        return False

# 1. El grafo no dirigido de la Sección 1
grafo_nd = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E'],
}

# 2. Un ciclo de 5 nodos (cada nodo se conecta con el anterior y el siguiente)
ciclo_5 = {
    '1': ['2', '5'],
    '2': ['1', '3'],
    '3': ['2', '4'],
    '4': ['3', '5'],
    '5': ['4', '1']
}

# 3. Puentes de Königsberg: 4 nodos (A=tierra norte, B=isla, C=tierra sur, D=tierra este)
konigsberg = {
    'A': ['B', 'B', 'D'],     # dos puentes a B, uno a D
    'B': ['A', 'A', 'C', 'C', 'D'],
    'C': ['B', 'B', 'D'],
    'D': ['A', 'B', 'C'],
}

print(f"Grafo Sección 1 tiene camino euleriano: {tiene_camino_euleriano(grafo_nd)}")
print(f"Ciclo de 5 nodos tiene camino euleriano: {tiene_camino_euleriano(ciclo_5)}")
print(f"Puentes de Königsberg tiene camino euleriano: {tiene_camino_euleriano(konigsberg)}")
````


### Ejercicio 3: Densidad y búsqueda

Genera grafos Erdős–Rényi con $n = 50$ nodos y probabilidad $p \in \{0.05, 0.1, 0.2, 0.4, 0.6, 0.8\}$.

2. La longitud del camino más corto

Grafica ambas métricas en función de $p$. ¿Qué patrón observas?


````python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
import random

# Definimos los parámetros del experimento
n_nodos = 50
probabilidades = [0.05, 0.1, 0.2, 0.4, 0.6, 0.8]
experimentos_por_p = 50  # Promediamos para eliminar el ruido estadístico

promedio_expandidos = []
promedio_longitud = []

def bfs_metrica(G, inicio, meta):
    """
    Implementación rigurosa de BFS basada en el Algoritmo Genérico.
    Cuenta exactamente cuántos nodos se extraen de la frontera.
    """
    cola = [inicio]
    visitados = {inicio}
    padre = {inicio: None}
    expandidos = 0

    while cola:
        actual = cola.pop(0)
        expandidos += 1

        #Prueba de meta
        if actual == meta:
            # Reconstruimos el camino para medir la longitud (d)
            longitud = 0
            recorrido = actual
            while padre[recorrido] is not None:
                longitud += 1   
                recorrido = padre[recorrido]
            return expandidos, longitud

        #Expandir vecinos
        for vecino in G.neighbors(actual):
            if vecino not in visitados:
                visitados.add(vecino)
                padre[vecino] = actual
                cola.append(vecino)

    return expandidos, -1 # Falla de seguridad (no debería ocurrir si están conectados)

# Ejecutamos las simulaciones
random.seed(42)

for p in probabilidades:
    exp_temp = []
    lon_temp = []
    
    for _ in range(experimentos_por_p):
        G = nx.erdos_renyi_graph(n_nodos, p)
        
        # Rigor: Para evitar grafos desconectados (islas) en p bajos, 
        # forzamos a elegir nodos del clúster más grande.
        componente_gigante = max(nx.connected_components(G), key=len)
        if len(componente_gigante) < 2:
            continue # Saltamos si el grafo es un conjunto de nodos aislados
            
        u, v = random.sample(list(componente_gigante), 2)
        
        expandidos, longitud = bfs_metrica(G, u, v)
        exp_temp.append(expandidos)
        lon_temp.append(longitud)
        
    promedio_expandidos.append(np.mean(exp_temp))
    promedio_longitud.append(np.mean(lon_temp))

# --- GRAFICAMOS LOS RESULTADOS ---
fig, ax1 = plt.subplots(figsize=(9, 5))

# Eje Y principal (Longitud del camino)
color = 'tab:red'
ax1.set_xlabel('Probabilidad de arista p (Densidad)')
ax1.set_ylabel('Longitud del camino más corto (d)', color=color)
ax1.plot(probabilidades, promedio_longitud, marker='o', color=color, linewidth=2, label='Longitud (d)')
ax1.tick_params(axis='y', labelcolor=color)

# Eje Y secundario (Nodos Expandidos)
ax2 = ax1.twinx()  
color = 'tab:blue'
ax2.set_ylabel('Nodos Expandidos', color=color)
ax2.plot(probabilidades, promedio_expandidos, marker='s', linestyle='--', color=color, linewidth=2, label='Expandidos')
ax2.tick_params(axis='y', labelcolor=color)

plt.title('BFS en Grafos Erdős–Rényi: Longitud vs Nodos Expandidos')
fig.tight_layout() 
plt.show()
````

Conclusión del Experimento:

1. Observación de la Gráfica
Al analizar las curvas generadas por las simulaciones en los grafos Erdős-Rényi, destacan dos comportamientos contrastantes:

    1) Colapso de la Distancia (Curva Roja): A medida que la probabilidad de conexión p (la densidad del grafo) aumenta, la longitud del camino más corto d decae drásticamente, acercándose rápidamente a 1 o 2 saltos.

    2) Sobrecarga de Expansión (Curva Azul): Contrario a la intuición básica, un camino más corto no significa menos trabajo. El número de nodos expandidos se mantiene alto (o incluso experimenta un pico en densidades medias) antes de descender solo cuando el grafo es tan denso que casi cualquier par de nodos está conectado directamente.

2. Interpretación Teórica
Este fenómeno ilustra perfectamente la tensión matemática en la complejidad temporal y espacial de la Búsqueda en Amplitud (BFS), dictada por la fórmula $\mathcal{O}(b^d)$:

En grafos dispersos (p bajo): La profundidad d es mayor, pero el factor de ramificación b (vecinos por nodo) es muy pequeño. La frontera crece lentamente.

En grafos densos (p medio-alto): Ocurre el efecto de "Mundo Pequeño" (Small-World). Los atajos masivos colapsan la profundidad d. Sin embargo, el factor de ramificación b explota linealmente con p.

Al inundar la cola FIFO con una cantidad masiva de vecinos desde el Nivel 1, BFS se ve obligado a procesar y expandir a casi toda la red antes de encontrar la meta en el Nivel 2. El algoritmo cambia profundidad por anchura extrema.

3. Implicaciones
Este experimento demuestra empíricamente el mayor punto débil de BFS: la densidad del grafo es el enemigo natural de la memoria.
Aunque BFS garantiza matemáticamente el camino óptimo (en número de aristas), en espacios de estado con alta conectividad (alta densidad), la frontera crece de forma inmanejable desde los primeros niveles. Esto justifica por qué en problemas del mundo real fuertemente conectados (como redes sociales densas, routing complejo o juegos de tablero), la búsqueda ciega a lo ancho es inviable por el desbordamiento de la RAM, haciendo estrictamente necesario el uso de Búsqueda Informada (Heurísticas) o estrategias como IDDFS para contener la explosión de la frontera.
